## Boolean Indexing 

````python
a = 1; b = 2;
if (a == 1) & (b == 2):
    print('T & T')
if (a == 1) & (b == 0):
    print('T & F')
if (a == 0) & (b == 2):
    print('F & T')
if (a == 0) & (b == 0):
    print('F & F')

if (a == 1) | (b == 2):
    print('T | T')
if (a == 1) | (b == 0):
    print('T | F')
if (a == 0) | (b == 2):
    print('F | T')
if (a == 0) | (b == 0):
    print('F | F')
````

## Database-Style DataFrame Joins

จะทำการทดสอบว่า การ join สองแบบได้ผลลัพธ์ตรงกัน โดยจะเอาผลลัพธ์มาต่อกันให้อยู่ใน data frame เดียวเพื่อให้ง่ายต่อการดู

````python
left = pd.DataFrame({'key1': ['foo', 'foo', 'bar'],
                     'key2': ['one', 'two', 'one'],
                     'lval': [1, 2, 3]})
right = pd.DataFrame({'key1': ['foo', 'foo', 'bar', 'bar'],
                      'key2': ['one', 'one', 'one', 'two'],
                      'rval': [4, 5, 6, 7]})
m1 = pd.merge(left, right, on=['key1', 'key2'], how='outer')
````

````python
left = pd.DataFrame({'lkey1': ['foo', 'foo', 'bar'],
                     'lkey2': ['one', 'two', 'one'],
                     'lval': [1, 2, 3]})
right = pd.DataFrame({'rkey1': ['foo', 'foo', 'bar', 'bar'],
                      'rkey2': ['one', 'one', 'one', 'two'],
                      'rval': [4, 5, 6, 7]})
m2 = pd.merge(left, right, left_on=['lkey1', 'lkey2'], right_on=['rkey1', 'rkey2'], how='outer')
````

````python
lkey = m2.loc[:, ['lkey1','lkey2']]
rkey = m2.loc[:, ['rkey1','rkey2']]
keypart = np.where(lkey.isnull(), rkey, lkey)
valpart = m2[['lval','rval']]
m2_derived = pd.DataFrame(np.hstack((keypart, valpart)), columns=['key1', 'key2', 'lval', 'rval'])
m2_derived
````

````python
m1.join(m2_derived.add_prefix('m2_'))
````

- ในกรณี 1 (grouping across column) aggregate function เวลาเรียกมันเอา input มาจาก ทุก row ในคราเดียว และมัน expect สิ่งที่ return เป็น vector ที่มี dimension เหมือน input

# 10.3 Revisit

ใน section 10.3 ข้างบน เราได้เห็นว่า ถ้า เรา split a Series vs. a DataFrame เราได้ผลลัพธ์ต่างกัน
- Split a Series: ค่าใน dict ถูกแปลงเป็น หลายๆ row ใน Series ใหญ่
- Split a DataFrame: dict ยังคงเป็น dict และหนึ่ง dict ถูกเก็บไว้ใน 1 row ของ Series ใหญ่ with dtype: object
    - เข้าใจว่าเพราะคำตอบมี column เดียว ผลลัพธ์เลยเป็น Series ใหญ่ไม่ใช่ DataFrame

ใน Section revisit นี้ เราพบอีกข้อแตกต่างคื
- Split a Series: ไม่ว่า callable จะ return dict or Series จะได้ ผลลัพธ์เดียวกัน
- Split a DataFrame: ถ้า callable return Series มันจะกลายเป็น 1 row ของ DataFrame ใหญ่
