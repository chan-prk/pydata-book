## Basic Indexing and Slicing

```python
import numpy as np
```

```python
arr = np.arange(10)
arr
arr[5]
arr[5:8]

#As you can see, if you assign a scalar value to a slice, as in arr[5:8] = 12, the value is
#propagated (or broadcasted henceforth) to the entire selection.
arr[5:8] = 12
arr
```

```python
l = list(range(10))
l[5:8] = [12]
l
```

## Boolean Indexing 

```python
a = 1; b = 2;
if (a == 1) & (b == 2):
    print('T & T')
if (a == 1) & (b == 0):
    print('T & F')
if (a == 0) & (b == 2):
    print('F & T')
if (a == 0) & (b == 0):
    print('F & F')

if (a == 1) | (b == 2):
    print('T | T')
if (a == 1) | (b == 0):
    print('T | F')
if (a == 0) | (b == 2):
    print('F | T')
if (a == 0) | (b == 0):
    print('F | F')
```

```python
# ข้อควรระวังในการใช้ --> ต้องใส่วงเล็บ
if a == 9 & a == 1:
    print('F & T')
if  (a == 9) & (a == 1):
    print('(F) & (T)')
```

```python
np.array([True, False]) | True
```

```python
np.array([True, False]) | False
```

```python
np.logical_and(np.array([True, False]), False)
```

## Filtering Out Missing Data

```python
data = pd.DataFrame([[1., 6.5, 3.], [1., NA, NA],
                     [NA, NA, NA], [NA, 6.5, 3.]])
data[4] = NA
```

```python
data
```

```python
data.dropna(axis=1, how='all')
```

จะ reimplement ให้เหมือน cell ข้างบนโดยใช้ numpy.all

```python
data_arr = data.values
flag = pd.isnull(data).values
mask = ~flag.all(axis=0)

ans01 = data_arr[:, mask]
print(ans01)

ans02 = data.loc[:, mask]
ans02
```
