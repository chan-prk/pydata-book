## Boolean Indexing 

```python
a = 1; b = 2;
if (a == 1) & (b == 2):
    print('T & T')
if (a == 1) & (b == 0):
    print('T & F')
if (a == 0) & (b == 2):
    print('F & T')
if (a == 0) & (b == 0):
    print('F & F')

if (a == 1) | (b == 2):
    print('T | T')
if (a == 1) | (b == 0):
    print('T | F')
if (a == 0) | (b == 2):
    print('F | T')
if (a == 0) | (b == 0):
    print('F | F')
```

```python
# ข้อควรระวังในการใช้ --> ต้องใส่วงเล็บ
if a == 9 & a == 1:
    print('F & T')
if  (a == 9) & (a == 1):
    print('(F) & (T)')
```

```python
np.array([True, False]) | True
```

```python
np.array([True, False]) | False
```

```python
np.logical_and(np.array([True, False]), False)
```

## Filling In Missing Data

จะทำการ fill missing value ด้วย mean*100 เวลา data frame มีหลาย columns

```python
np.random.seed(12345)

col1 = np.random.randn(10, 1)*0.5 + 5
col2 = np.random.randn(10, 1)*0.1 + 3
cols = np.hstack((col1, col2))
# cols[[4, 9]][:, [0, 1]] = np.nan # ถ้าใช้แบบนี้ original cols ไม่ถูก affected
cols[[4, 9]] = np.nan

data = pd.DataFrame(cols, columns=['X1', 'X2'])
mseries = data.apply(lambda x: x.mean(), axis=0)
mseries *= 100

data.fillna(mseries.to_dict())
```

```python
mnames = ['movie_id', 'title', 'genres']
movies = pd.read_table('../datasets/movielens/movies.dat', sep='::',
                       header=None, names=mnames)
movies[:3]
```

```python
all_genres = []
for x in movies.genres:
    all_genres.extend(x.split('|'))
genres = pd.unique(all_genres)
genres
```

```python
#scratchpad
np.where(genres == 'Comedy')
```

```python
#scratchpad
a1 = np.array([2])
a2 = np.array([4, 8, 16])
a3 = np.array([32, 64])
a1
```

```python
#scratchpad
np.concatenate([a1, a2, a3])
```

```python
#scratchpad
l = ['Comedy', 'Fantasy', 'Romance']
li = [np.where(genres == e)[0] for e in l] # np.where คืน result เป็น tuple of array เราต้องแกะออกมาหนึ่งชั้นให้เป็น array เฉยๆ
np.concatenate(li)
```

```python
def findIndex(l):
    return np.concatenate([np.where(genres == e)[0] for e in l])
```

```python
#scratchpad
findIndex('Comedy|Romance'.split('|'))
```

### บททดสอบ 1

```python
zero_matrix = np.zeros((len(movies), len(genres)))
dummies = pd.DataFrame(zero_matrix, columns=genres)
```

```python
for i, gen in enumerate(movies.genres):
    indices = dummies.columns.get_indexer(gen.split('|')) # วิธีจากหนังสือ
    my_indices = findIndex(gen.split('|'))
    if i < 5:
        print(f"indices:{indices}\nmy_indices:{my_indices}")
    if not np.array_equal(indices, my_indices):
        print('Different indices found!')
        break
```
