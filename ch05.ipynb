### DataFrame

The DataFrame has both a row and column index; it can be thought of
as a dict of Series all sharing the same index.

There are many ways to construct a DataFrame, though one of the most common is
from a dict of equal-length lists or NumPy arrays

````python
data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada', 'Nevada'],
        'year': [2000, 2001, 2002, 2001, 2002, 2003],
        'pop': [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}
frame = pd.DataFrame(data)
````

````python
frame
````

````python
frame.head()
````

#### Operations between DataFrame and Series

````python
arr = np.arange(12.).reshape((3, 4))
print(arr)
print(arr[0]) #can be broadcast to have shape == 1x4
arr - arr[0]
````

````python
frame = pd.DataFrame(np.arange(12.).reshape((4, 3)),
                     columns=list('bde'),
                     index=['Utah', 'Ohio', 'Texas', 'Oregon'])
series = frame.iloc[0]
print(frame)
print(series)
````

By default, arithmetic between DataFrame and Series matches the index of the Series
on the DataFrame’s columns, **broadcasting down the rows**. ( serie ถูก pad ให้มี 1 row จากนั้น มันได้ถูกขยายไปเป็นหลายๆ rows)

````python
#default
# Cell (X)
frame - series
````

````python
#ลองเอง
mean_series = pd.Series({'e': 6.5, 'b': 4.5, 'd': 5.5})
frame - mean_series
````

If an index value is not found in either the DataFrame’s columns or the Series’s index,
the objects will be reindexed to form the union:

````python
series2 = pd.Series(range(3), index=['b', 'e', 'f'])
frame + series2
````

If you want to instead broadcast over the columns (จากเดิม `series3` มีแค่ 1 column ถูกขยายจนมี จ.น. column เท่ากับ `frame`), matching on the rows, you have to
use one of the arithmetic methods. The axis number that you pass is the axis to match on. In this case we mean to match
on the DataFrame’s row index (`axis='index'` or `axis=0`) and broadcast across.

ก็คือ การ duplicate in columns direction จะเกิดขึ้นได้ series ก็จะต้องถูกปรับให้เป็น column vector หลังจากนั้นมันก็จะถูกเอาไปบวก กับ column vector ที่มาจากแต่ละ column ของ frame (column vector ไม่สามารถบวกกับ row vector ได้ ดังนั้น serie จะต้องบวกกับ column เท่านั้น) และเนื่องจากแต่ละ column ใน frame ถูก label ด้วย index หลายๆ index การที่เราจะเอา series ไปบวกกับมัน เราจะต้อง match index

````python
series3 = frame['d']
print(frame)
print(series3)
print(frame.sub(series3, axis='index'))
print(frame.sub(series3)) 
# By default, arithmetic between DataFrame and Series matches the index of the Series on the DataFrame’s columns
````

````python
frame = pd.DataFrame(np.random.randn(4, 3), columns=list('bde'),
                     index=['Utah', 'Ohio', 'Texas', 'Oregon'])
frame
np.abs(frame)
print(frame)
````

````python
f = lambda x: x.max()
frame.apply(f)
````

>Here the function `f`, which computes the maximum of a Series, is invoked once on each column in `frame`. 
**The result is a Series having the columns of `frame` as its index.**

ประโยคสุดท้ายตีความได้ว่า columns are retained

````python
#ลองเอง #default
frame.apply(lambda x: x.max(), axis='index') 
# from doc: 0 or 'index': apply function to each column.
# invoke once per column --> the computation is performed across the whole index (or all rows) of "each column"
# Each value of the result belongs to each column and so it is labeld with the column name --> "columns are retained"
````

````python
frame.apply?
````

If you pass `axis='columns'` to apply, the function will be invoked once per row
instead:

````python
frame.apply(lambda x: x.max(), axis='columns')
# The computation is performed across all columns of "each row". 
# Each element of an output belongs to each row. We label it using the row's index label. --> rows are retained
````

````python
print(frame.max(axis='index'), '\n')
print(frame.max(axis='columns'))
````

````python
def f(x):
    return pd.Series([x.min(), x.max()], index=['min', 'max'])
frame.apply(f)
# Per each column, each serie is returned.
#series #return
````

เพื่อทำความเข้าใจประโยคในหนังสือ
>by default rank
breaks ties by assigning each group the mean rank

> Here, instead of using the average rank 6.5 for the entries 0 and 2, they instead have
been set to 6 and 7 because label 0 precedes label 2 in the data.

จึงเขียน R code นี้ขึ้นมา
```
library(tidyverse)

df <- tibble(
  data=c(7, -5, 7, 4, 2, 0, 4),
  rank_method_first=c(6., 1., 7., 4., 3., 2., 5.)
  ) |> mutate(data = as.integer(data))

df

df |> group_by(data) |> 
      mutate(mean_rank = mean(rank_method_first)) |>
      print() |>
      mutate(mean_rank_format = scales::label_number(0.1)(mean_rank)) |>
      ungroup() |>
      select(last_col())

###### Output ######
  mean_rank_format
  <chr>           
1 6.5             
2 1.0             
3 6.5             
4 4.5             
5 3.0             
6 2.0             
7 4.5
```

````python
obj = pd.Series([7, -5, 7, 4, 2, 0, 4])
obj.rank()
````

````python
obj.rank(method='first')
````

>You can rank in descending order, too:

```
df1 <- tibble(
  data=c(7, -5, 7, 4, 2, 0, 4),
  method_first_desc=c(1., 7., 2., 3., 5., 6., 4.)
) |> mutate(data = as.integer(data))

df1 |> group_by(data) |> 
  mutate(max_rank = max(method_first_desc)) |>
  print() |>
  mutate(max_rank_format = scales::label_number(0.1)(max_rank)) |>
  ungroup() |>
  select(last_col())
  
###### Output ######
# A tibble: 7 × 1
  max_rank_format
  <chr>          
1 2.0            
2 7.0            
3 2.0            
4 4.0            
5 5.0            
6 6.0            
7 4.0
```

````python
# Assign tie values the maximum rank in the group
print(obj.rank(ascending=False, method='first'))
obj.rank(ascending=False, method='max')
````

```
import pandas_datareader.data as web
all_data = {ticker: web.get_data_yahoo(ticker)
            for ticker in ['AAPL', 'IBM', 'MSFT', 'GOOG']}

price = pd.DataFrame({ticker: data['Adj Close']
                     for ticker, data in all_data.items()})

                      for ticker, data in all_data.items()})
```

````python
returns = price.pct_change()
returns.tail()
````

The `corr` method of Series computes the correlation of the overlapping, non-NA,
aligned-by-index values in two Series. Relatedly, `cov` computes the covariance:

````python
returns['MSFT'].corr(returns['IBM'])
````

````python
returns['MSFT'].cov(returns['IBM'])
````

````python
returns.MSFT.corr(returns.IBM)
````

````python
print(type(returns.MSFT))
print(type(returns))
````

````python
returns.corr()
returns.cov()
````

Using DataFrame’s `corrwith` method, you can compute pairwise correlations
between a DataFrame’s columns or rows with another Series or DataFrame. Passing a
Series returns a Series with the correlation value computed for each column:

````python
print(returns.corrwith(returns.IBM), '\n')

print(returns.IBM.corr(returns.AAPL))
print(returns.IBM.corr(returns.GOOG))
print(returns.IBM.corr(returns.MSFT))
````

Passing a DataFrame computes the correlations of **matching column names**. Here I
compute correlations of percent changes with volume:

````python
print(returns.corrwith(volume),'\n')

print(returns.AAPL.corr(volume.AAPL))
print(returns.GOOG.corr(volume.GOOG))
print(returns.IBM.corr(volume.IBM))
print(returns.MSFT.corr(volume.MSFT))
````
