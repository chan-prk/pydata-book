## Functions

````python
def my_function(x, y, z=1.5):
    if z > 1:
        return z * (x + y)
    else:
        return z / (x + y)
````

Each function can have positional arguments and keyword arguments. Keyword arguments
are most commonly used to specify default values or optional arguments. In
the preceding function, x and y are positional arguments while z is a keyword argument.
This means that the function can be called in any of these ways:
```
my_function(5, 6, z=0.7)
my_function(3.14, 7, 3.5)
my_function(10, 20)
```

The main restriction on function arguments is that the keyword arguments must follow
the positional arguments (if any). You can specify keyword arguments in any
order; this frees you from having to remember which order the function arguments
were specified in and only what their names are.

````python
def your_function(x=1.5, y , z):
    return True
````

It is possible to use keywords for passing positional arguments as
well. In the preceding example, we could also have written:
```
my_function(x=5, y=6, z=7)
my_function(y=6, x=5, z=7)
```
In some cases this can help with readability.

````python
my_function(z=5, y=6, x=7)
````

````python
my_function(z=5, y=6, 7)
````

### Currying: Partial Argument Application

partial สร้างได้จาก closure ที่มา [SO](https://stackoverflow.com/a/37056348/718529)

````python
def partial(f, *args):
    def pf(*rest):
        return f(*args, *rest)
    return pf
````

````python
def add_numbers(x, y):
    return x + y
add_five = partial(add_numbers, 5)
````

````python
add_five(4)
````

### Generators

Having a consistent way to iterate over sequences, like objects in a list or lines in a
file, is an important Python feature. This is accomplished by means of the iterator
protocol, a generic way to make objects iterable. For example, iterating over a dict
yields the dict keys:

````python
some_dict = {'a': 1, 'b':2, 'c':3}
````

````python
for key in some_dict:
    print(key)
````

When you write `for key in some_dict`, the Python interpreter first attempts to create
an iterator out of `some_dict`:

````python
dict_iterator = iter(some_dict)
dict_iterator
````

Most methods expecting a list or list-like object will also
accept any iterable object. This includes built-in methods such as `min, max,` and `sum`,
and type constructors like `list` and `tuple`:

````python
list(dict_iterator)
````

A generator is a concise way to construct a new iterable object.

To create
a generator, use the `yield` keyword instead of `return` in a function:

````python
def squares(n = 10):
    print(f"Generating square numbers from 1 to {n ** 2}")
    for x in range(1, n+1):
        yield x ** 2
````

````python
gen = squares()
gen
````

````python
for x in gen:
    print(x, end=' ')
````

````python
for x in gen:
    print(x, end=' ')
````

### Charactor/byte Counting

````python
f1 = open(path, mode = 'rb')
print(f1.tell())
f1.read(10)
````

The `read` method advances the file handle’s position by the number of bytes read.
`tell` gives you the current position:

````python
f1.tell()
````

#search

เริ่มที่ node 1 ถ้าอ่านข้อมูล n ครั้ง ก็ต้องข้าม(advance)ไป n hop แต่ว่าถ้าข้ามแค่ n-1 hop ก็จะมาหยุดที่ node n (node n อยู่ห่างจาก node 1 เป็นระยะ n-1 hop) ดังนั้นถ้าข้าม n hop ตัว position ก็จะอยู่ที่ node n+1

````python
f2 = open(path, encoding = 'utf-8')
print(f2.tell())
f2.read(10)
````

````python
f2.tell()
````

ในกรณีนี้ มีการอ่าน ไป 10 charactors แต่ว่า ñ กินไป 2 bytes ก็เลย
1. final position มากขึ้นหนึ่ง เพราะว่า advance เป็นจำนวน bytes มากกว่า
2. อ่านถึง charactor ที่อยู่ไกลกว่าซึ่งก็คือ r เพราะ ว่า ตอนที่เจอ bytes ยาวๆอย่าง \xc3\xb1 มันถูกนับเป็น 1 แทนที่จะเป็น 2 ทำให้ quota ของ charactors ที่เหลือ(จนกว่าจะครบ 10) เหลือมากกว่า ก็เลยรับ charactor เข้ามาในเซตของตัวที่ถูกอ่านมากกว่า

````python
f1.close(); f2.close()
````

### Bytes and Unicode with Files

````python
sink_path = 'sink.txt'
with open(path) as source:
    with open(sink_path, 'wt', encoding='iso-8859-1') as sink:
        sink.write(source.read())

with open(sink_path, encoding='iso-8859-1') as f:
    print(f.read(10))
````

````python
sink_path = 'sink.txt'
with open(path, encoding='utf-8') as source: # บรรทัดนี้คือตัวปัญหา ถ้าไม่ใส่ encoding มันชอบ read ด้วย tis-620
    with open(sink_path, 'wt', encoding='utf-8') as sink:
        sink.write(source.read())

with open(sink_path, encoding='utf-8') as f:
    print(f.read(10))
````
